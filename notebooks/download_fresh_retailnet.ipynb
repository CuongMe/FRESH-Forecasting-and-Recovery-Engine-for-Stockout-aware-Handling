# Download FreshRetailNet datasets

Downloads both Hugging Face repositories into `../data/`. Re-running is safe: unchanged files are reused from the Hugging Face cache.

```python
%pip install -q -U huggingface_hub
```

```python
from pathlib import Path
import os
from huggingface_hub import snapshot_download

DATA_DIR = Path('../data').resolve()
DATASETS = {
    'FreshRetailNet-50K': 'Dingdong-Inc/FreshRetailNet-50K',
    'FreshRetailNet-LT': 'Dingdong-Inc/FreshRetailNet-LT',
}
HF_TOKEN = os.getenv('HF_TOKEN') or os.getenv('HUGGINGFACE_HUB_TOKEN')
DATA_DIR.mkdir(parents=True, exist_ok=True)
print(f'Downloading to: {DATA_DIR}')
```

```python
for name, repo_id in DATASETS.items():
    folder = DATA_DIR / name
    snapshot_download(
        repo_id=repo_id,
        repo_type='dataset',
        local_dir=folder,
        token=HF_TOKEN,
        ignore_patterns=['.gitattributes'],
    )
    assert any(folder.rglob('*')), f'No files downloaded for {name}'
    print(f'✓ {name}: {folder}')
```

For private or gated datasets, set `HF_TOKEN` before launching Jupyter or run `huggingface-cli login`.
